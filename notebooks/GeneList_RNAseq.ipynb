PEAK OVERLAPS between ATF4 data and the Basal and Stress induced conditions in Liver (26960794)

```python
import pandas as pd
rna_seq = pd.read_table('ATF4_RNAseq/WT_and_KO_Tmc_treatment_GSE76771/GSE76771_ALL_gene_expression.txt', header=4)
rna_seq
```

```python
sig_rna = rna_seq[rna_seq['significant'] == 'yes']
sig_rna = sig_rna[sig_rna['gene']!= '']
sig_rna = sig_rna[sig_rna['gene']!= '-']
sig_rna
```

```python
# Filter the rows where the 'gene' column contains 'Mar' or 'Sep'
date_genes = sig_rna[sig_rna['gene'].str.contains(r'\b(Mar|Sep)\b', na=False, regex=True)]

# Extract the 'gene' column as a list
date_gene_list = date_genes['gene'].tolist()

print(date_gene_list)
```

```python
# Mapping of month abbreviations to full names
month_mapping = {
    'Mar': 'March','Sep': 'Sept'
}

# Function to rename genes
def rename_gene(gene_name):
    if isinstance(gene_name, str):
        parts = gene_name.split('-', 1)  # Split only on the first '-'
        if len(parts) == 2:
            number, month_abbr = parts
            month_full = month_mapping.get(month_abbr, month_abbr)  # Use the mapping or keep the abbreviation
            return f"{month_full}{number}"
    return gene_name  # Return the original name if not in the 'number-Month' format or if not a string

# Apply the function to the 'gene' column
sig_rna['gene'] = sig_rna['gene'].apply(rename_gene)

# Display the updated DataFrame
print(sig_rna)
```

```python
# Filter the DataFrame for rows where 'sample_1' is 'q1' and 'sample_2' is 'q2'
filtered_df = sig_rna[(sig_rna['sample_1'] == 'q3') & (sig_rna['sample_2'] == 'q4')]

filtered_df
```

```python
# Remove entries with NaN in the 'gene' column
sig_rna_cleaned = filtered_df.dropna(subset=['gene'])

# Display the cleaned DataFrame
sig_rna_cleaned
```

```python
sig_rna_cleaned['gene']
```

```python
# Use your annotated (.txt) file here
atf4_aad = pd.read_table('Mito_stress_treatment/Amino_acid_deprivation/ATF4/results/annotation/ATF4_AAD_annotated.txt')
atf4_aad['SYMBOL']
```

```python
genes_to_check = sig_rna_cleaned['gene'].tolist()
filtered_genes_new = atf4_aad[atf4_aad['SYMBOL'].isin(sig_rna_cleaned['gene'])]
filtered_genes_new
```

```python
filtered_genes_new.to_csv('ATF4_RNAseq/WT_and_KO_Tmc_treatment_GSE76771/ER_overlaps/ATF4_ER_peaks_overlap_with_KO_vs_KO-TT.bed', sep = '\t')
```

PEAK OVERLAPS between ATF4 data and the Multiomics paper (28566324)

```python
# Load the gene list
with open('ATF4_RNAseq/Multiomics/Multiomics_DE_genes.txt') as f:
    genes = set(line.strip().lower() for line in f)

# Check the genes in the file and save to a .bed file
with open('Mito_stress_treatment/ER_stress/ATF4/results/annotation/ATF4_WT_Tmc_vs_ATF4_KO_Tmc_annotated.txt') as f, open('ATF4_RNAseq/Multiomics/ATF4_ER_overlaps_with_DEgenes_Multiomics.bed', 'w') as out_file:
    for line in f:
        columns = line.strip().split('\t')
        if columns[22].lower() in genes:  # Assuming gene names are in the second column
            out_file.write(line)
```

```python
import pandas as pd

# Load the list of genes from the gene file
gene_file = 'Alzheimers.txt'
with open(gene_file, 'r') as f:
    gene_list = set(line.strip() for line in f)

# Load the data file into a pandas DataFrame
data_file = 'HMD_HumanPhenotype.csv'
data_df = pd.read_csv(data_file, sep=',', header=None)

# Filter the DataFrame for rows where the gene name in the first column matches the gene list
matched_df = data_df[data_df[0].isin(gene_list)]

# Select the first and third columns from the matched rows
result_df = matched_df[[2]]

# Save the result to a new file
result_file = 'matched_genes_alzheimers.tsv'
result_df.to_csv(result_file, sep='\t', index=False, header=False)

print(f'Matched genes and their third column details saved to {result_file}')
```
