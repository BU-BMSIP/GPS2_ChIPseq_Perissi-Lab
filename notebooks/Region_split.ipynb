```python
import pandas as pd
import re

# Define your annotation terms
terms_to_match = {
    'exon': 'Exon',
    'intron': 'Intron',
    'distal_intergenic': 'Distal Intergenic',
    '3UTR': "3' UTR",
    'promoter': 'Promoter',
    'downstream': 'Downstream',
    '5UTR': "5' UTR"
}

# Function to match annotation and return the matched term
def match_annotation(annotation):
    for key, pattern in terms_to_match.items():
        if re.search(pattern, annotation, re.IGNORECASE):
            return key
    return None

# Path to your annotation file
annotation_file = "Treatment/ATF4/results/annotation/ATF4_ISR_annotated.txt"

# Read annotation file into pandas DataFrame
df = pd.read_table(annotation_file)

# Apply pattern matching to create a new column 'annotation_type'
df['annotation_type'] = df['annotation'].apply(match_annotation)

# Filter rows based on matched annotation types and save to .bed files
for key in terms_to_match.keys():
    filtered_df = df[df['annotation_type'] == key]
    if not filtered_df.empty:
        output_file = f'Treatment/ATF4/results/annotation/ATF4_ISR/{key}_annotations.bed'
        filtered_df.to_csv(output_file, sep='\t', columns=['seqnames', 'start', 'end'], index=False, header=False)
        print(f'Saved {key} annotations to {output_file}')
    else:
        print(f'No {key} annotations found.')

print('All annotation types processed.')
```
